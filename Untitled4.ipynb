### 3. Melatih Model YOLO
Karena dataset *Paddy Disease Classification* umumnya merupakan tugas klasifikasi gambar (Image Classification), kita dapat menggunakan model klasifikasi YOLO (misalnya `yolov8n-cls.pt` atau `yolo11n-cls.pt`).

Di bawah ini adalah kode untuk memulai training klasifikasi gambar menggunakan Ultralytics YOLO.

```python
import os
import glob
import random
import shutil
import zipfile
from ultralytics import YOLO
import torch

zip_path = '/content/paddy-disease-classification.zip'
raw_data_path = '/content/paddy_data/train_images'
dataset_root = '/content/paddy_dataset'
train_dest = os.path.join(dataset_root, 'train')
val_dest = os.path.join(dataset_root, 'val')

# 1. Ekstrak dataset jika file zip ada dan folder data mentah masih kosong
if os.path.exists(zip_path):
    if not os.path.exists(raw_data_path) or len(os.listdir(raw_data_path)) == 0:
        print("Mengekstrak dataset padi dari file ZIP lokal...")
        os.makedirs('/content/paddy_data', exist_ok=True)
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall('/content/paddy_data')
        print("Ekstraksi selesai!")
else:
    print(f"Peringatan: File zip {zip_path} tidak ditemukan.")

# 2. Menyiapkan Struktur Direktori untuk YOLO
if os.path.exists(raw_data_path) and len(os.listdir(raw_data_path)) > 0:
    # Jika folder tujuan belum terisi, lakukan pembagian data train & val
    if not os.path.exists(train_dest) or len(os.listdir(train_dest)) == 0:
        print("Membagi dataset menjadi 80% train dan 20% val...")
        os.makedirs(train_dest, exist_ok=True)
        os.makedirs(val_dest, exist_ok=True)

        classes = [d for d in os.listdir(raw_data_path) if os.path.isdir(os.path.join(raw_data_path, d))]

        for cls in classes:
            os.makedirs(os.path.join(train_dest, cls), exist_ok=True)
            os.makedirs(os.path.join(val_dest, cls), exist_ok=True)

            images = glob.glob(os.path.join(raw_data_path, cls, '*'))
            random.shuffle(images)

            split_idx = int(len(images) * 0.8)
            train_imgs = images[:split_idx]
            val_imgs = images[split_idx:]

            for img in train_imgs:
                shutil.copy(img, os.path.join(train_dest, cls, os.path.basename(img)))
            for img in val_imgs:
                shutil.copy(img, os.path.join(val_dest, cls, os.path.basename(img)))

        print("Pembagian dataset selesai!")
    else:
        print("Dataset terstruktur train/val sudah siap.")
else:
    print(f"Error: Folder train_images mentah tidak ditemukan di {raw_data_path}.")

# 3. Inisialisasi Model YOLOv8
model = YOLO('yolov8n-cls.pt')

# Mendeteksi device (GPU jika tersedia, jika tidak menggunakan CPU)
device = 0 if torch.cuda.is_available() else 'cpu'
print(f"Menggunakan device: {device}")

# 4. Mulai Training jika folder sudah siap dan tidak kosong
if os.path.exists(train_dest) and len(os.listdir(train_dest)) > 0:
    results = model.train(
        data=dataset_root,
        epochs=10,
        imgsz=224,
        batch=32,
        device=device
    )
else:
    print("Error: Proses persiapan dataset gagal. Silakan periksa kembali folder dataset Anda.")
```

### 4. Uji Coba Model (Inference / Prediksi)
Setelah proses pelatihan di atas selesai, Anda dapat menjalankan kode di bawah ini untuk menguji model dengan gambar penyakit padi baru (misalnya dari folder `test_images`).

```python
from ultralytics import YOLO
import os
import glob
from PIL import Image
import matplotlib.pyplot as plt

# 1. Muat model terbaik hasil training terbaru (train-5)
model_path = '/content/runs/classify/train-5/weights/best.pt'

if os.path.exists(model_path):
    model_test = YOLO(model_path)

    # 2. Ambil sampel beberapa gambar dari folder test_images
    test_images_path = glob.glob('/content/paddy_data/test_images/*')

    if test_images_path:
        # Ambil 3 contoh gambar secara acak untuk prediksi
        import random
        sample_images = random.sample(test_images_path, min(3, len(test_images_path)))

        for img_path in sample_images:
            # Lakukan prediksi
            results = model_test(img_path)

            # Tampilkan hasil prediksi
            for result in results:
                # Dapatkan nama kelas dengan probabilitas tertinggi
                top_class_idx = result.probs.top1
                top_class_name = result.names[top_class_idx]
                top_class_conf = float(result.probs.top1conf) * 100

                # Tampilkan Gambar dan Hasilnya
                img = Image.open(img_path)
                plt.figure(figsize=(5, 5))
                plt.imshow(img)
                plt.title(f"Prediksi: {top_class_name} ({top_class_conf:.2f}%)")
                plt.axis('off')
                plt.show()
    else:
        print("Folder test_images kosong atau tidak ditemukan.")
else:
    print(f"Model di {model_path} belum ditemukan. Harap pastikan folder run pelatihan Anda benar.")
```
